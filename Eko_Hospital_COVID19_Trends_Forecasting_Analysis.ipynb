```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('/Users/mdshahriar/Downloads/Python Projects/Project 3/covid19_trends_data.csv')

# Convert the Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Plotting the trends of cumulative cases, deaths, and recoveries
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['Cumulative_Cases'], label="Cumulative Cases", color='blue')
plt.plot(df['Date'], df['Cumulative_Deaths'], label="Cumulative Deaths", color='red')
plt.plot(df['Date'], df['Cumulative_Recoveries'], label="Cumulative Recoveries", color='green')
plt.xlabel("Date")
plt.ylabel("Count")
plt.title("Cumulative COVID-19 Trends Over Time")
plt.legend()
plt.grid(True)
plt.show()
```

### Code Explanation:

- **Import Libraries:**

  - `import matplotlib.pyplot as plt`: Imports Matplotlib to create visualizations (graphs and plots).

- **Load the Dataset:**

- **Convert the Date Column:**
  - `df['Date'] = pd.to_datetime(df['Date'])`: Converts the 'Date' column to the `datetime` format for easier manipulation and plotting over time.

- **Plot the Data:**

  - `plt.plot(df['Date'], df['Cumulative_Cases'], label="Cumulative Cases", color='blue')`: Plots the 'Cumulative Cases' data with a blue line.

# Assessing Testing Effectiveness

```python
# Correlation between New Cases and New Tests
correlation = df['New_Cases'].corr(df['New_Tests'])
print(f"Correlation between new cases and new tests: {correlation}")

# Plotting the relationship between new cases and new tests
plt.figure(figsize=(10, 6))
plt.scatter(df['New_Tests'], df['New_Cases'], alpha=0.5, color='purple')
plt.title("New Cases vs. New Tests")
plt.xlabel("New Tests Conducted")
plt.ylabel("New Cases")
plt.grid(True)
plt.show()
```

  - `print(f"Correlation between new cases and new tests: {correlation}")`: Prints the correlation value to the console.

- **Plot the Relationship:**
  - `plt.figure(figsize=(10, 6))`: Sets the figure size of the plot to 10x6 inches.
  - `plt.scatter(df['New_Tests'], df['New_Cases'], alpha=0.5, color='purple')`: Creates a scatter plot with 'New_Tests' on the x-axis and 'New_Cases' on the y-axis. The `alpha=0.5` makes the points semi-transparent, and the color is set to purple.

# Analyzing Mortality Rates

```python
# Calculate mortality rate
df['Mortality_Rate'] = (df['Cumulative_Deaths'] / df['Cumulative_Cases']) * 100

# Plotting mortality rate over time
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['Mortality_Rate'], label="Mortality Rate (%)", color='red')
plt.xlabel("Date")
plt.ylabel("Mortality Rate (%)")
plt.title("Mortality Rate Over Time")
plt.grid(True)
plt.legend()
plt.show()
```

### Code Explanation:

- **Calculate Mortality Rate:**

It divides the 'Cumulative_Deaths' by 'Cumulative_Cases' and multiplies by 100 to get the rate as a percentage.

- **Plot the Mortality Rate:**
  - `plt.figure(figsize=(12, 6))`: Sets the figure size of the plot to 12x6 inches.
  - `plt.plot(df['Date'], df['Mortality_Rate'], label="Mortality Rate (%)", color='red')`: Plots the mortality rate over time, using the 'Date' for the x-axis and 'Mortality_Rate' for the y-axis. The line is colored red.
  - `plt.xlabel("Date")`: Labels the x-axis as "Date".
  - `plt.ylabel("Mortality Rate (%)")`: Labels the y-axis as "Mortality Rate (%)".
  - `plt.title("Mortality Rate Over Time")`: Sets the title of the plot.
  - `plt.grid(True)`: Adds a grid to the plot for better readability.
  - `plt.legend()`: Displays a legend to label the line as "Mortality Rate (%)".
  - `plt.show()`: Displays the plot.

# Tracking Recovery Patterns

```python
# Calculate recovery rate (Recovery Rate = Cumulative Recoveries / Cumulative Cases)
df['Recovery_Rate'] = (df['Cumulative_Recoveries'] / df['Cumulative_Cases']) * 100

# Plotting recovery rate over time
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['Recovery_Rate'], label="Recovery Rate (%)", color='green')
plt.xlabel("Date")
plt.ylabel("Recovery Rate (%)")
plt.title("Recovery Rate Over Time")
plt.grid(True)
plt.legend()
plt.show()
```

# Forecasting Future Cases with ARIMA

```python
from statsmodels.tsa.arima.model import ARIMA
import warnings
warnings.filterwarnings('ignore')


# Set the date as index for time series forecasting
df.set_index('Date', inplace=True)

# Fit an ARIMA model on Cumulative Cases
model = ARIMA(df['Cumulative_Cases'], order=(5, 1, 0))  # ARIMA(p, d, q) parameters
model_fit = model.fit()

# Predict the next 30 days
forecast = model_fit.forecast(steps=30)

# Plotting the forecast
plt.figure(figsize=(12, 6))
plt.plot(df.index, df['Cumulative_Cases'], label="Historical Data")
plt.plot(pd.date_range(df.index[-1], periods=30, freq='D'), forecast, label="Forecasted Cases", color='orange')
plt.xlabel("Date")
plt.ylabel("Cumulative Cases")
plt.title("COVID-19 Cumulative Cases Forecast")
plt.legend()
plt.grid(True)
plt.show()
```

### Code Explanation:

- **Import Libraries:**
  - `from statsmodels.tsa.arima.model import ARIMA`: Imports the ARIMA model from the `statsmodels` library for time series forecasting.
  - `import warnings`: Imports the warnings module to suppress warnings.
  - `warnings.filterwarnings('ignore')`: Ignores warnings to keep the output clean.

- **Set Date as Index:**
  - `df.set_index('Date', inplace=True)`: Sets the 'Date' column as the index for the DataFrame, which is necessary for time series analysis.

- **Fit the ARIMA Model:**
  - `model = ARIMA(df['Cumulative_Cases'], order=(5, 1, 0))`: Initializes the ARIMA model with the 'Cumulative_Cases' data. The parameters `(5, 1, 0)` represent:
    - `p=5`: The number of lag observations included in the model (AR part).
    - `d=1`: The number of times the data needs to be differenced to make it stationary (I part).
    - `q=0`: The number of lagged forecast errors in the prediction equation (MA part).
  - `model_fit = model.fit()`: Fits the ARIMA model to the data.

- **Make Forecast:**
  - `forecast = model_fit.forecast(steps=30)`: Forecasts the next 30 days of cumulative cases using the fitted ARIMA model.

- **Plot the Forecast:**
  - `plt.figure(figsize=(12, 6))`: Sets the figure size of the plot to 12x6 inches.
  - `plt.plot(df.index, df['Cumulative_Cases'], label="Historical Data")`: Plots the historical cumulative cases.
  - `plt.plot(pd.date_range(df.index[-1], periods=30, freq='D'), forecast, label="Forecasted Cases", color='orange')`: Plots the forecasted cumulative cases for the next 30 days, using the last date in the data and a frequency of 'D' (daily).
  - `plt.xlabel("Date")`: Labels the x-axis as "Date".
  - `plt.ylabel("Cumulative Cases")`: Labels the y-axis as "Cumulative Cases".
  - `plt.title("COVID-19 Cumulative Cases Forecast")`: Sets the title of the plot.
  - `plt.legend()`: Displays a legend to differentiate between historical and forecasted data.
  - `plt.grid(True)`: Adds a grid to the plot for better readability.
  - `plt.show()`: Displays the plot.

# Evaluating Policy Impact

```python
# Example: Evaluate the impact of a specific policy on new cases
# Let's assume lockdown started on "2020-04-01" for analysis purposes.
policy_start_date = pd.to_datetime("2020-04-01")
pre_policy_data = df[df.index < policy_start_date]
post_policy_data = df[df.index >= policy_start_date]

# Compare the average number of new cases before and after the policy
pre_policy_avg = pre_policy_data['New_Cases'].mean()
post_policy_avg = post_policy_data['New_Cases'].mean()

print(f"Average new cases before policy: {pre_policy_avg}")
print(f"Average new cases after policy: {post_policy_avg}")
```

### Code Explanation:

- **Define Policy Start Date:**
  - `policy_start_date = pd.to_datetime("2020-04-01")`: Defines the start date of the policy (lockdown) as "2020-04-01" and converts it into a datetime object.

- **Split Data into Pre-Policy and Post-Policy:**
  - `pre_policy_data = df[df.index < policy_start_date]`: Filters the dataset to include only the data before the policy start date (i.e., before April 1, 2020).
  - `post_policy_data = df[df.index >= policy_start_date]`: Filters the dataset to include only the data after the policy start date (i.e., from April 1, 2020, onward).

- **Calculate Average New Cases:**
  - `pre_policy_avg = pre_policy_data['New_Cases'].mean()`: Calculates the average number of new cases before the policy by computing the mean of the 'New_Cases' column in the pre-policy data.
  - `post_policy_avg = post_policy_data['New_Cases'].mean()`: Calculates the average number of new cases after the policy by computing the mean of the 'New_Cases' column in the post-policy data.

- **Print the Results:**
  - `print(f"Average new cases before policy: {pre_policy_avg}")`: Prints the average number of new cases before the policy.
  - `print(f"Average new cases after policy: {post_policy_avg}")`: Prints the average number of new cases after the policy.
